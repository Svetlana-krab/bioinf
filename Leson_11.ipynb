```python
#11.1
numbers=input("Введите числа через пробел")
input_list=[int(x) for x in numbers.split()]
from numbers import Number
def get_min(input_list: list) -> list:
    assert isinstance(input_list, list), "Не список"
    assert len(input_list)>0, "Пустой список"

    min_value= input_list[0]
    indexes=[0]
    for idx, i in enumerate(input_list):
        assert isinstance(i, (int, float, complex)) and not isinstance(i, bool), "Не только числа"
        if i<min_value:
            min_value=i
            indexes=[idx]
        elif i==min_value:
            indexes.append(idx)

    return indexes
result=get_min(input_list)
print(f"Минимальное значение: {result}")
print(f"Индексы значений: {input_list[result[0]]}")
```

```python
#11.2
def do_nothing_decorator(func):
    return func

@do_nothing_decorator
def bubble_sorted(a: list) -> list:
    assert isinstance(a,list), "Не список"
    assert len(a)>0, "Пустой список"

    for i in a:
        assert isinstance(i, (int,float, complex)) and not isinstance(i,bool), "Не только числа"

    arr =a.copy()

    n=len(arr)
    for i in range(n-1):
        flag=False
        for j in range(n-1-i):
            if arr[j]>arr[j+1]:
                arr[j],arr[j+1]=arr[j+1],arr[j]
                flag=True
        if not flag:
            break

    return arr
    
numbers=input("Введите числа через пробел: ")
input_list = [int(x) for x in numbers.split()]

sorted_list=bubble_sorted(input_list)
print(f"Исходный список: {input_list}")
print(f"Отсортированный: {sorted_list}")
```

```python
#11.3 
import random

length = int(input("ВВедите длину последовательности"))

def generator_dna(length: int) -> str:

    assert isinstance(length, int), "Длина должна быть целым числом"
    assert length>0, "Длина должна быть положительной"

    nucleotides= "ACGT"
    posled=''.join(random.choices(nucleotides, k=length))
    return posled

dna=generator_dna(length)
print(f"Сгенерированная ДНК: {dna}")
```

```python
#Fibbonacci Numbers
def fibonacci(n:int) -> int:
    assert isinstance(n, int), "n должно быть целым числом"
    assert n>=0, "n должно быть >=0"
    assert n<=25, "n должно быть <=25"

    if n==0:
        return 0
    elif n ==1:
        return 1
    
    a,b=0,1
    for i in range(2, n+1):
        a,b=b,a+b
    return b

n=int(input("Введите n(0-25):"))
result=fibonacci(n)
print(f"F({n}) = {result}")
```

```python
#Degree Array
def count_degrees(n, edges):
    assert isinstance(n,int) and n>0, "n должна быть положительным целым числом"
    assert isinstance(edges, list), "edges должен быть списком"
    
    degrees = [0]*(n+1)
    for a, b in edges:
        degrees[a]+=1
        degrees[b]+=1
    return degrees[1:]
first_line=input().split()
n=int(first_line[0])
m=int(first_line[1])

edges=[]
for i in range(m):
    a,b = map(int,input().split())
    edges.append((a,b))
result=count_degrees(n,edges)

for i, degree in enumerate(result,start=1):
    print(f"К точке {i} ведут {degree} дорог(и)")
```

```python
#Insertion Sort
def count_swaps(n, arr):
    swaps=0

    for i in range(1,n):
        j=i
        while  j>0 and arr[j]<arr[j-1]:
            arr[j],arr[j-1]=arr[j-1],arr[j]
            swaps+=1
            j-=1
    return swaps

n=int(input())
arr=list(map(int, input().split()))
result=count_swaps(n,arr)
print(result)
```

```python
#Counting DNA Nucleotides
def count_nucleotides(dna: str) -> dict:

    k={'A':0, 'C':0, 'G':0, 'T':0}
    for nucleotide in dna:
        assert nucleotide in 'ACGT', f"Символ {nucleotide} не является нуклеотидом"

    for nucleotide in dna:
        k[nucleotide]+=1
    return k

dna=input("Ввдеите последовательность ДНК:").upper()
result=count_nucleotides(dna)
print(f"A {result['A']}")
print(f"C {result['C']}")
print(f"G {result['G']}")
print(f"T {result['T']}")
```

```python
#Transcribing DNA into RNA
def transcrib_dna_to_rna(dna:str)->str:
    for nucleotide in dna:
        assert nucleotide in 'ACGT', f"Символ {nucleotide} не является нуклеотидом"
    rna=dna.replace('T','U')
    return rna
dna=input("Ввдеите последовательность ДНК:").upper()
rna=transcrib_dna_to_rna(dna)
print(f"ДНК: {dna}")
print(f"РНК: {rna}")
```

```python
#Complementing a Strand of DNA
def revers_complement(dna:str)->str:
    for nucleotide in dna:
        assert nucleotide in 'ACGT', f"Символ {nucleotide} не является нуклеотидом"
    complement={'A':'T', 'T':'A', 'C':'G', 'G':'C', }
    result=''
    for nucleotide in reversed(dna):
        result +=complement[nucleotide]
    return result
dna=input("Ввдеите последовательность ДНК:").upper()
result=revers_complement(dna)
print(f"ДНК: {dna}")
print(f"Результат: {result}")
```

```python
#Rabbits and Recurrence Relations
def rabbit_pairs(n:int,k:int)->int:
    if n==1:
        return 1
    if n==2:
        return 1

    prev_prev=1
    prev=1

    for month in range(3,n+1):
        current=prev +k*prev_prev  #i-предыдущее значение
        prev_prev=prev
        prev=current
    return prev

print("Введите n и k через пробел:")
n,k=map(int,input().split())
result=rabbit_pairs(n,k)
print(f"Через {n} месяцев будет {result} пар кроликов")

    
```
