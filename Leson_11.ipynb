```python
#11.1
numbers=input("Введите числа через пробел")
input_list=[int(x) for x in numbers.split()]
from numbers import Number
def get_min(input_list: list) -> list:
    assert isinstance(input_list, list), "Не список"
    assert len(input_list)>0, "Пустой список"

    min_value= input_list[0]
    indexes=[0]
    for idx, i in enumerate(input_list):
        assert isinstance(i, (int, float, complex)) and not isinstance(i, bool), "Не только числа"
        if i<min_value:
            min_value=i
            indexes=[idx]
        elif i==min_value:
            indexes.append(idx)

    return indexes
result=get_min(input_list)
print(f"Минимальное значение: {result}")
print(f"Индексы значений: {input_list[result[0]]}")
```

```python
#11.2
def do_nothing_decorator(func):
    return func

@do_nothing_decorator
def bubble_sorted(a: list) -> list:
    assert isinstance(a,list), "Не список"
    assert len(a)>0, "Пустой список"

    for i in a:
        assert isinstance(i, (int,float, complex)) and not isinstance(i,bool), "Не только числа"

    arr =a.copy()

    n=len(arr)
    for i in range(n-1):
        flag=False
        for j in range(n-1-i):
            if arr[j]>arr[j+1]:
                arr[j],arr[j+1]=arr[j+1],arr[j]
                flag=True
        if not flag:
            break

    return arr
    
numbers=input("Введите числа через пробел: ")
input_list = [int(x) for x in numbers.split()]

sorted_list=bubble_sorted(input_list)
print(f"Исходный список: {input_list}")
print(f"Отсортированный: {sorted_list}")
```

```python
#11.3 
import random

length = int(input("ВВедите длину последовательности"))

def generator_dna(length: int) -> str:

    assert isinstance(length, int), "Длина должна быть целым числом"
    assert length>0, "Длина должна быть положительной"

    nucleotides= "ACGT"
    posled=''.join(random.choices(nucleotides, k=length))
    return posled

dna=generator_dna(length)
print(f"Сгенерированная ДНК: {dna}")
```

```python
#11.4
import random

def guess_number():
    secret_number = random.randint(1,100)
    print("Игра в \"Угадай число\"")
    print("У Вас 10 попыток")

    for attempt in range(1,11):
        print(f"\nПопытка {attempt} из 10")
        guess = int(input("ваше число"))
        if guess == secret_number:
            print(f"Вы угадали число {secret_number} за {attempt} попыток! Поздравляю!")
            break
        elif guess < secret_number:
            print(f"Загаданное число БОЛЬШЕ {guess}")
        elif guess > secret_number:
            print(f"Загаданное число МЕНЬШЕ {guess}")
    else:
        print(f"Попытки закончились! Загаданное число {secret_number}")

guess_number()
```
